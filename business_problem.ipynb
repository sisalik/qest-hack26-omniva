**Differnt variations of the classical TSP**

Omniva needs to solve the TSP for their shipments every day. There are always a number of packages that need to be delivered that day, there is a set time window when things need to be delievered and the number of curiers needs to be minimised.

The problem will be solved in steps:
1. Solve the classical TSP on 5 or more vertices. We try to find the maximum number vertices. Making sure that all team members understand what is going on. Use only one car - only one route.
2. Add a second car to the route and try to minimise the distance.
3. Add a time constraint to the routes. Seems like the edges will simply have multiple weights and the sum of the time weights mustn't be larger than the threshold.
